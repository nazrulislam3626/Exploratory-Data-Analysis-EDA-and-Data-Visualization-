```python
import pandas as pd

def get_all_value_counts(dataframe, columns):
    all_counts = []
    for col in columns:
        # Ensure the column exists and is not entirely unique (like 'ID')
        if col in dataframe.columns and dataframe[col].nunique() < len(dataframe):
            counts = dataframe[col].value_counts().reset_index()
            counts.columns = ['Value', 'Count']
            counts['Feature'] = col
            all_counts.append(counts)

    if all_counts:
        return pd.concat(all_counts, ignore_index=True)
    else:
        return pd.DataFrame(columns=['Feature', 'Value', 'Count'])

# List of columns to tabulate value counts for
columns_for_counts = ['Family', 'Education', 'Personal Loan', 'Securities Account', 'CD Account', 'Online', 'CreditCard']

# Get the consolidated value counts
consolidated_counts_df = get_all_value_counts(df, columns_for_counts)

# Display the combined table
display(consolidated_counts_df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Get unique features from the consolidated DataFrame
unique_features = consolidated_counts_df['Feature'].unique()

# Plotting each feature's value counts
for feature in unique_features:
    # Filter data for the current feature
    feature_df = consolidated_counts_df[consolidated_counts_df['Feature'] == feature]

    plt.figure(figsize=(8, 5))
    sns.barplot(x='Value', y='Count', data=feature_df, palette='viridis')
    plt.title(f'Value Counts for {feature}')
    plt.xlabel(feature)
    plt.ylabel('Count')
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate skewness and kurtosis for numerical columns
skewness_values = df.skew(numeric_only=True)
kurtosis_values = df.kurt(numeric_only=True)

# Create DataFrames for plotting
skew_df = skewness_values.reset_index()
skew_df.columns = ['Feature', 'Skewness']

kurt_df = kurtosis_values.reset_index()
kurt_df.columns = ['Feature', 'Kurtosis']

# Plotting skewness and kurtosis in subplots
plt.figure(figsize=(18, 7))

# Skewness Plot
plt.subplot(1, 2, 1) # 1 row, 2 columns, first plot
sns.barplot(x='Feature', y='Skewness', data=skew_df, palette='coolwarm')
plt.title('Skewness of Numerical Features')
plt.xlabel('Features')
plt.ylabel('Skewness')
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.7)

# Kurtosis Plot
plt.subplot(1, 2, 2) # 1 row, 2 columns, second plot
sns.barplot(x='Feature', y='Kurtosis', data=kurt_df, palette='viridis')
plt.title('Kurtosis of Numerical Features')
plt.xlabel('Features')
plt.ylabel('Kurtosis')
plt.xticks(rotation=90)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()
plt.show()
```

### Skewness Analysis:
Skewness measures the asymmetry of the probability distribution of a real-valued random variable about its mean. A negative skew indicates the tail on the left side of the distribution is longer or fatter, while a positive skew indicates that the tail on the right side is longer or fatter.

*   **Features with low skewness (close to 0):** `ID`, `Age`, `Experience`, `Family`, `Education`, `Online`. These variables exhibit a relatively symmetrical distribution, meaning their values are spread out fairly evenly around the mean.
*   **Features with positive skewness (right-skewed):** `Income`, `CCAvg`, `Mortgage`, `Personal Loan`, `Securities Account`, `CD Account`, `CreditCard`. This indicates that these features have a longer tail towards higher values. For example, `Income`, `CCAvg`, and `Mortgage` might have a few very high values that pull the mean to the right of the median. `Personal Loan`, `Securities Account`, and `CD Account` are binary variables with a large majority of 0s and a small number of 1s, which inherently leads to high positive skewness. `CreditCard` also shows positive skew, suggesting more customers do not have a credit card compared to those who do.
*   **Features with negative skewness (left-skewed):** `ZIP Code`. The extremely high negative skewness for `ZIP Code` suggests a long tail towards lower values. However, `ZIP Code` is typically a categorical identifier rather than a numerical value with a meaningful distribution in this context, so its skewness value should be interpreted with caution.

### Kurtosis Analysis:
Kurtosis measures the 'tailedness' of the probability distribution of a real-valued random variable. High kurtosis means more outliers or extreme values, leading to a sharper peak and fatter tails (leptokurtic). Low kurtosis means fewer outliers, leading to a flatter peak and lighter tails (platykurtic).

*   **Features with low/negative kurtosis (platykurtic):** `ID`, `Age`, `Experience`, `Family`, `Education`, `Online`, `CreditCard`. These features have lighter tails than a normal distribution, implying fewer extreme outliers compared to a normal distribution.
*   **Features with high kurtosis (leptokurtic):** `Income`, `CCAvg`, `Mortgage`, `Personal Loan`, `Securities Account`, `CD Account`, `ZIP Code`. These variables have heavier tails, indicating a higher likelihood of extreme values or outliers. `CD Account` shows very high kurtosis, suggesting a very concentrated distribution around its mean with very few occurrences of extreme values (which are the 1s). `Mortgage` and `CCAvg` also show significant positive kurtosis, pointing to the presence of some exceptionally high values in these features. Similar to skewness, `ZIP Code`'s extremely high kurtosis is likely due to its nature as an identifier.

First, let's create age groups by binning the `Age` column. This will allow us to analyze income levels for different age brackets.

```python
# Define age bins and labels
age_bins = [20, 30, 40, 50, 60, 70]
age_labels = ['20-29', '30-39', '40-49', '50-59', '60-69']

# Create a new column 'Age Group'
df['Age Group'] = pd.cut(df['Age'], bins=age_bins, labels=age_labels, right=False)

# Display the first few rows with the new 'Age Group' column
display(df[['Age', 'Age Group', 'Income']].head())
```

Now, let's calculate the mean income for each age group and visualize it with a bar plot.

```python
# Calculate the mean income for each age group
age_income = df.groupby('Age Group')['Income'].mean().reset_index()

# Create the bar plot
plt.figure(figsize=(10, 6))
sns.barplot(x='Age Group', y='Income', data=age_income, palette='viridis')
plt.title('Mean Income by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Mean Income')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
